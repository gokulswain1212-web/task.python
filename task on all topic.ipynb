## 14. Create class BankAccount with private attribute __balance. Add deposit and withdrawmethods with validation.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposited:", amount)
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print("Withdrawn:", amount)

    def show_balance(self):
        print("Balance:", self.__balance)


account = BankAccount(1000)

account.show_balance()

account.deposit(500)
account.show_balance()

account.withdraw(300)
account.show_balance()

account.withdraw(2000)
```

## 18. Create class Rectangle with private length and width. Add getter, setter and method tocalculate area.

```python
class Rectangle:
    def __init__(self, length, width):
        self.__length = length  
        self.__width = width 
        
    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width


    def set_length(self, length):
        if length > 0:
            self.__length = length
        else:
            print("Invalid length")

    def set_width(self, width):
        if width > 0:
            self.__width = width
        else:
            print("Invalid width")

    def calculate_area(self):
        return self.__length * self.__width


rect = Rectangle(10, 5)

print("Length:", rect.get_length())
print("Width:", rect.get_width())
print("Area:", rect.calculate_area())
rect.set_length(20)
rect.set_width(10)

```

## 22. Create multiple inheritance with two parent classes Father and Mother, and child classChild.

```python
class Father:
    def __init__(self, father_name, father_job):
        self.FatherName = father_name
        self.FatherJob = father_job


class Mother:
    def __init__(self, mother_name, mother_hobby):
        self.MotherName = mother_name
        self.MotherHobby = mother_hobby


class Child(Father,Mother):
    def __init__(self, father_name, father_job, mother_name, mother_hobby, child_name, age):
        Father.__init__(self, father_name, father_job)
        Mother.__init__(self, mother_name, mother_hobby)
        self.ChildName = child_name
        self.Age = age

    def display_info(self):
        print("Father Name:", self.FatherName)
        print("Father Job:", self.FatherJob)
        print("Mother Name:", self.MotherName)
        print("Mother Hobby:", self.MotherHobby)
        print("Child Name:", self.ChildName)
        print("Child Age:", self.Age)


c = Child("Ramesh", "Engineer", "Sita", "Painting", "Gokul", 10)

c.display_info()
```

## 23. Create abstract class Shape with abstract method area(). Implement it in Circle andRectangle using super().

```python
from abc import ABC , abstractmethod

class Shape(ABC):
    def __init__(self,name):
        self.Name=name

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, name, radius):
        super().__init__(name)
        self.Radius = radius

    def area(self):
        area=3.14*self.Radius*self.Radius
        print("area of circle:",area)

class Rectangle(Shape):
    def __init__(self, name, length, width):
        super().__init__(name)
        self.Length = length
        self.Width = width

    def area(self):
        area=self.Length*self.Width
        print("area of Rectangle:",area)

circle=Circle("circle",8)
circle.area()
rectangle=Rectangle("Rectangle",12,5)
rectangle.area()
```

## 24. Create hybrid inheritance with classes A, B, C and D (D inherits from B and C).

```python
class A:
    def __init__(self, a_data, **kwargs):
        super().__init__(**kwargs)
        self.AData = a_data


class B(A):
    def __init__(self, a_data, b_data, **kwargs):
        super().__init__(a_data, **kwargs)
        self.BData = b_data


class C(A):
    def __init__(self, a_data, c_data, **kwargs):
        super().__init__(a_data, **kwargs)
        self.CData = c_data


class D(B, C):
    def __init__(self, a_data, b_data, c_data, d_data):
        super().__init__(a_data, b_data, c_data=c_data)
        self.DData = d_data

    def display(self):
        print("Data from A:", self.AData)
        print("Data from B:", self.BData)
        print("Data from C:", self.CData)
        print("Data from D:", self.DData)


obj = D("A Class Data", "B Class Data", "C Class Data", "D Class Data")

obj.display()
```

## 27. Create class Point and overload + operator to add two points.

```python
class Point:
    def __init__(self, x, y):
        self.X = x
        self.Y = y

    def __add__(self, other):
        return Point(self.X + other.X, self.Y + other.Y)

    def display(self):
        print("Point:", self.X, self.Y)


p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2
p3.display()
```

## 30. Create class ComplexNumber and overload both + and - operators.

```python
class ComplexNumber:
    def __init__(self, real, imaginary):
        self.Real = real
        self.Imaginary = imaginary

    def __add__(self, other):
        return ComplexNumber(
            self.Real + other.Real,
            self.Imaginary + other.Imaginary
        )

    def __sub__(self, other):
        return ComplexNumber(
            self.Real - other.Real,
            self.Imaginary - other.Imaginary
        )

    def display(self):
        print(self.Real, "+", self.Imaginary, "i")


c1 = ComplexNumber(5, 3)
c2 = ComplexNumber(2, 1)

add_result = c1 + c2
sub_result = c1 - c2

print("Addition:")
add_result.display()

print("Subtraction:")
sub_result.display()
```

```python
start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

count = 0

print("Prime numbers are:")

for num in range(start, end + 1):
    if num < 2:
        continue   

    for i in range(2, num):
        if num % i == 0:
            break   
    else:
        print(num)
        count += 1

print("Total prime numbers:", count)
```
