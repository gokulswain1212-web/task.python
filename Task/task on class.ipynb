## 1. Create a class Student with instance attributes name and age. Create two objectsand print their details.

```python
class student:
    def __init__(self,name,age):
        self.Name=name
        self.Age=age

    def display_info(self):
        print("student name is ",self.Name)
        print("student age is ",self.Age)


s1=student("gokul",21)
s1.display_info()
print(s1.Name)
print(s1.Age)
```

## 2. Define class Car with attributes brand, model, and year. Create an object and display all attributes.

```python
class Car:
    def __init__(self,brand,model,year):
        self.Brand=brand
        self.Model=model
        self.Year=year

    def display_car(self):
        print("your car brand is ",self.Brand)
        print("car model is",self.Model)
        print("in the year",self.Year)

c=Car("thar","sxd",1990)

c.display_car()
```
