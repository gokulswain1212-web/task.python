```python
sum = 0
count = 0

while True:
    num = int(input("Enter a number: "))

    if num < 0:
        break

    sum += num
    count += 1

if count > 0:
    average = sum / count
    print("Sum =", sum)
    print("Average =", average)
else:
    print("No positive numbers were entered.")
```

# 5. Find the factorial of a number using for loop and also using while loop.

```python
# for loop
num = int(input("Enter a number: "))

factorial = 1

for i in range(1, num + 1):
    factorial *= i

print("Factorial =", factorial)
```

```python
# while loop
num = int(input("Enter a number: "))

factorial = 1
i = 1

while i <= num:
    factorial *= i
    i += 1

print("Factorial =", factorial)
```

```python
## 16. Create multi-level inheritance: Person → Employee → Manager. Override a
method at each level.
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    


class Employee(Person):
    def __init__(self, name, age, emp_id, department):
        super().__init__(name, age)
        self.emp_id = emp_id
        self.department = department

   

class Manager(Employee):
    def __init__(self, name, age, emp_id, department, team_size):
        super().__init__(name, age, emp_id, department)
        self.team_size = team_size

    def display_info(self):
        print("Manager Details:")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Employee ID:", self.emp_id)
        print("Department:", self.department)
        print("Team Size:", self.team_size)



m = Manager("gokul", 35, "M201", "AI Department", 10)
m.display_info()
```

```python
## 17. Create hierarchical inheritance: Parent Animal → Dog and Cat. Override sound()
method.
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print(self.name, "says: Woof Woof")


class Cat(Animal):
    def sound(self):
        print(self.name, "says: Meow Meow")



dog = Dog("dogesh")
dog.sound()
print()
cat = Cat("biladon")
cat.sound()
```

```python
## 18. Create abstract class Shape with abstract method area(). Implement it in Circle
and Square using super().
```

```python
from abc import ABC, abstractmethod
import math

#
class Shape(ABC):
    def __init__(self, value):
        self.value = value

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        super().__init__(radius)   # Calls Shape constructo

    def area(self):
        return math.pi * self.value ** 2


# Square class
class Square(Shape):
    def __init__(self, side):
        super().__init__(side)     # Calls Shape constructor

    def area(self):
        return self.value ** 2


# Main program
c = Circle(7)
s = Square(5)

print("Area of Circle =", c.area())
print("Area of Square =", s.area())
```
