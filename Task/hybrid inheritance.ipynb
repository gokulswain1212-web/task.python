```python
## Hybrid Inheritance Questions (1-10)
1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits
from both B and C).
Concept
Hybrid Inheritance (Combination of Single + Multiple Inheritance)
What to do
• Create parent class A.
• Create class B inheriting from A.
• Create another class C.
• Create class D inheriting from both B and C.
How to solve
1. Create class A.
2. Add one method.
3. Create class B(A).
4. Create class C.
5. Create class D(B, C).
6. Create an object of D.
7. Call methods from all parent classes.
```

```python
class A:
    def show_a(self):
        print("Class A: Student Name = Rahul")
class B(A):
    def show_b(self):
        print("Class B: Roll Number = 101")

class C:
    def show_c(self):
        print("Class C: Department = Computer Science")

class D(B, C):
    def show_d(self):
        print("Class D: Hybrid Inheritance Example")
obj = D()

obj.show_a()   
obj.show_b() 
obj.show_c()  
obj.show_d()  
    
```

```python
## 2. Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
Concept
Real-world Hybrid Inheritance
What to do
• Person is the base class.
• Employee inherits Person.
• Student inherits Person.
• TeachingAssistant inherits Employee and Student.
How to solve

Hierarchy
Person
/ \
Employee Student
\ /
TeachingAssistant
Steps
1. Store name in Person.
2. Employee stores salary.
3. Student stores marks.
4. TeachingAssistant stores department.
5. Display all information.
```

```python
class Person:
    def __init__(self,name,**kwarge):
        super().__init__(**kwarge)
        self.Person_name=name

class Employee(Person):
    def __init__(self,emp_salary,**kwarge):
        super().__init__(**kwarge)
        self.Employee_salary=emp_salary

class Student(Person):
    def __init__(self,std_marks,**kwarge):
        super().__init__(**kwarge)
        self.Student_marks=std_marks

class TeachingAssistant(Employee,Student):
    def __init__(self,name,emp_salary,std_marks,department):
        super().__init__(
            name=name,
            emp_salary=emp_salary,
            std_marks=std_marks,
            #department=department,
        )
        self.TeachingAssistant_department=department

    def display_all_info(self):
        print("name:",self.Person_name)
        print("salary:",self.Employee_salary)
        print("marks:",self.Student_marks)
        print("department:",self.TeachingAssistant_department)



obj=TeachingAssistant("gokul",1000000,98,"python developer")
obj.display_all_info()

        
        
```

```python
## 3. Create Vehicle → Car, Electric → ElectricCar (ElectricCar inherits from both).
Concept
Hybrid Inheritance
What to do
• Vehicle is parent.
• Car inherits Vehicle.
• Electric is another parent.
• ElectricCar inherits Car and Electric.
How to solve
Hierarchy
Vehicle
|
Car Electric
\ /
ElectricCar
Steps
1. Vehicle has start().
2. Car has drive().
3. Electric has battery().
4. ElectricCar accesses all methods.
```

```python
class Vehicle:
    def __init__(self, brand,**kwarge):
        super().__init__(**kwarge)
        self.Brand = brand

    def start(self):
        print(f"{self.Brand} vehicle is starting.")


class Car(Vehicle):
    def __init__(self, model,**kwarge):
        super().__init__(**kwarge)
        self.Model = model

    def drive(self):
        print(f"{self.Brand} {self.Model} is being driven.")

class Electric:
    def __init__(self, battery_capacity,**kwarge):
        super().__init__(**kwarge)
        self.Battery_capacity = battery_capacity

    def battery(self):
        print(f"Battery Capacity: {self.Battery_capacity} kWh")

class ElectricCar(Car, Electric):
    def __init__(self, brand, model, battery_capacity):
        super().__init__(
            brand=brand,
            model=model,
            battery_capacity=battery_capacity
        )
        #self. Battery_capacity=battery_capacity

    def display(self):
        print("Electric Car Details")
        print("Brand:", self.Brand)
        print("Model:", self.Model)
        print("Battery:", self.Battery_capacity, "kWh")


ec = ElectricCar("Tesla", "Model 3", 75)

ec.display()
ec.start()
ec.drive()
ec.battery()
    
```

```python
## 4. Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO).
Concept
Method Resolution Order (MRO)
What to do
Create
A
|
B

C

D(B,C)
Print
D.mro()
How to solve
1. Create four classes.
2. Give same method name in multiple classes.
3. Print ClassName.mro().
4. Observe Python's searching order.
```

```python
class A:
    def message(self):
        print("Message from Class A")


class B(A):
    def message(self):
        print("Message from Class B")


class C(A):
    def message(self):
        print("Message from Class C")


class D(B, C):
    pass

obj = D()
obj.message()
print(D.mro())
```

```python
## 5. Create class A, B, C, D and solve the Diamond Problem using super().
Concept
Diamond Problem
Hierarchy
A
/ \
B C
\ /
D
What to do
Use super() everywhere.
How to solve
1. Create constructor in A.
2. Call super() in B.
3. Call super() in C.
4. Call super() in D.
5. Create object of D.
6. Observe A constructor runs only once.
```

```python
class A:
    def __init__(self,name,**kwarge):
        super().__init__(**kwarge)
        self.Name=name

class B(A):
    def __init__(self,age,**kwarge):
        super().__init__(**kwarge)
        self.Age=age

class C(A):
    def __init__(self,marks,**kwarge):
        super().__init__(**kwarge)
        self.Marks=marks

class D(B,C):
    def __init__(self,name,age,marks):
        super().__init__(
            name=name,
            age=age,
            marks=marks,
        )

    def display_info(self):
        print("name:",self.Name)
        print("age:",self.Age)
        print("marks:",self.Marks)


d=D("gokul",21,90)
d.display_info()
```

```python
## 6. Create a banking system using hybrid inheritance.
Concept
Hybrid + Real-world Example
What to do
Account
|
Savings

Loan

SavingsLoanAccount
How to solve
1. Account stores account number.
2. Savings stores balance.
3. Loan stores loan amount.
4. SavingsLoanAccount inherits both.
5. Print all details.
```

```python
class Account:
    def __init__(self, account_number, **kwargs):
        super().__init__(**kwargs)
        self.account_number = account_number

class Savings(Account):
    def __init__(self, balance, **kwargs):
        super().__init__(**kwargs)
        self.balance = balance


class Loan:
    def __init__(self, loan_amount, **kwargs):
        super().__init__(**kwargs)
        self.loan_amount = loan_amount


class SavingsLoanAccount(Savings, Loan):
    def __init__(self, account_number, balance, loan_amount):
        super().__init__(
            account_number=account_number,
            balance=balance,
            loan_amount=loan_amount
        )

    def display(self):
        print("Account Number :",self.account_number)
        print("Savings Balance:",self.balance)
        print("Loan Amount :",self.loan_amount)

obj = SavingsLoanAccount(1001, 50000, 200000)
obj.display()
```

```python
## 7. Create Shape, Color, Circle, and ColoredCircle.
Concept
Hybrid Inheritance
What to do
Shape
|
Circle

Color

ColoredCircle
How to solve
1. Shape has area().
2. Color has color().
3. Circle inherits Shape.
4. ColoredCircle inherits Circle and Color.
5. Display area and color.
```

```python

class Shape:
    def __init__(self, radius, **kwargs):
        super().__init__(**kwargs)
        self.radius = radius

    def area(self):
        area = 3.14 * self.radius * self.radius
        print("Area of Circle:", area)

class Circle(Shape):
    def __init__(self, **kwargs):
        super().__init__( **kwargs)

class Color:
    def __init__(self, color, **kwargs):
        super().__init__(**kwargs)
        self.color_name = color

    def color(self):
        print("Color:", self.color_name)


class ColoredCircle(Circle, Color):
    def __init__(self, radius, color):
        super().__init__(
            radius=radius,
            color=color
        )

    def display(self):
        print("Radius:", self.radius)
        self.area()
        self.color()


obj = ColoredCircle(7, "red")
obj.display()
    
```

```python
## 15. Create Circle and Rectangle both having area().
Concept
Duck Typing
What to do
Function
calculate(obj)
How to solve
1. Circle implements area().
2. Rectangle implements area().
3. Function calls area().
4. Print returned value.
```

```python
class Circle:
    def __init__(self,r):
        self.Redius=r

    def area(self):
        area=3.14*self.Redius*self.Redius
        print("area of circle :",area)

class Rectangle:
    def __init__(self,width,length):
        self.Width=width
        self.Length=length

    def area(self):
        area=self.Width*self.Length
        print("area of Rectangle",area)

def calculate(obj):
    obj.area()


c=Circle(8)
r=Rectangle(10,8)
calculate(c)
calculate(r)
```

```python
## 16. Create CreditCard and UPI both having pay().
Concept
Duck Typing
What to do
Payment system
How to solve
1. CreditCard → pay().
2. UPI → pay().
3. Function processes payment using pay().
```

```python

class CreditCard:
    def __init__(self, card_number, holder_name):
        self.card_number = card_number
        self.holder_name = holder_name

    def pay(self, amount):
        print(f"Payment of {amount} done using Credit Card")
        print(f"Card Holder: {self.holder_name}")
        print(f"Card Number: {self.card_number}")

class UPI:
    def __init__(self, upi_id):
        self.upi_id = upi_id

    def pay(self, amount):
        print(f"Payment of {amount} done using UPI")
        print(f"UPI ID: {self.upi_id}")

def process_payment(payment_method, amount):
    payment_method.pay(amount)  


credit_card = CreditCard("1234-5678-9012", "gokul")
upi = UPI("gokul@upi")


process_payment(credit_card, 500)
print("------------------")
process_payment(upi, 300)
```

```python
## 18. Create Logger classes using log(message).
Concept
Duck Typing
What to do
Two unrelated logger classes.
How to solve
1. FileLogger → log().
2. ConsoleLogger → log().
3. Function accepts logger object.
4. Calls log(message).
```

```python

class FileLogger:
    def log(self, message):
        print(f"Writing to file: {message}")

class ConsoleLogger:
    def log(self, message):
        print(f"Console: {message}")

def write_log(logger, message):
    logger.log(message)

file_logger = FileLogger()
console_logger = ConsoleLogger()

write_log(file_logger, "File log message")
write_log(console_logger, "Console log message")
```

```python
## 19. Create a hybrid inheritance structure and apply Duck Typing.
Concept
Hybrid + Duck Typing

Sumit Sir Notes
What to do
Create hybrid inheritance.
Create another unrelated class with same method.
How to solve
1. Create hybrid hierarchy.
2. Final child has method show().
3. Another unrelated class also has show().
4. One function calls show() for both.
```
