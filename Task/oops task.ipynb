## • brand
## • speed
## Method
## increase_speed(value)
## Maximum speed should never exceed 200 km/h.

```python
class car_speed:
    def car_info(self,brand,speed):
        self.brand=brand
        self.speed=speed


    def car_information(self):
        print("car brand is ",self.brand)
        print("car speed is ",self.speed)
    

    def increase_speed(self,value):
        if self.speed+value<=200:
            self.speed+=value
            print("speed increased successfuliy")

        else:
            self.speed=200
            print("maximum speed is 200 km/h")



        print("car brand",self.brand)
        print("current speed",self.speed)
        


car=car_speed()

car.car_info("thar",190)
car.car_information()
car.increase_speed(50)
car.increase_speed(50)
            
```

## 8. Movie Ticket
## Attributes
## • movie_name
## • seats
## Method
## book_ticket(number)
## Booking should happen only if seats are available.

```python
class Movie_ticket:
    def movie_info(self,movie_name,seats):
        self.movie_name=movie_name
        self.seats=seats


    def display_info(self):
         print("movie name is",self.movie_name)
         print("seats is",self.seats)
    

    def book_ticket(self,number):
        if number<=self.seats:
            self.seats -= number
            print("ok seat are avilable you can watch movie")

        else:
            print("sorry seats are not avilable")

    
    
movie=Movie_ticket()
movie.movie_info("game of throns",60)
movie.display_info()

movie.book_ticket(40)
movie.book_ticket(60)
movie.book_ticket(20)


```

## 10. Product
## Attributes
## • product_name
## • quantity
## Method
## sell(quantity)
## Quantity should never become negative.

```python
class Product:
    def product_info(self,product_name,quantity):
        self.product_name=product_name
        self.quantity=quantity



    def sell_quantity(self,sell):
        if sell<=self.quantity:
            self.quantity-=sell
            print("remaining quentity",self.quantity)

        else:
            print("not enougnh stock are available")


p= Product()
p.product_info("laptop",10)
p.sell_quantity(4)
p.sell_quantity(11)
```
