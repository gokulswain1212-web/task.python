## 15. (Medium) Take marks and print grade: ≥90 → A, ≥80 → B, ≥70 → C, else →fail

```python
mark=int(input("enter your mark:"))
if mark>=90:
     print("your grade is A",mark)
elif mark>=80:
     print("your grade is B",mark)
elif mark>=70:
     print("your grade is C",mark)
else:
    print("you fail")
```
