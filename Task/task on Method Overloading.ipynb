```python
## 1. Create a Calculator class that can add 2 numbers or 3 numbers using the same method
name.
Concept
Method Overloading (Python uses default arguments or *args to achieve overloading.)
What to do
• Create a class Calculator.
• Create one method add().
• The method should work for either 2 numbers or 3 numbers.
How to solve
1. Create class Calculator.
2. Define method add().
3. Use default arguments or *args.
4. Check how many numbers are provided.
5. Return the sum.
```

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

calc = Calculator()

print(calc.add(10, 20))      
print(calc.add(10, 20, 30))  
```

```python
## 2. Create an Area class that calculates the area of a square and rectangle using the same
method.
Concept
Method Overloading using default arguments.
What to do
• One parameter → Square.
• Two parameters → Rectangle.
How to solve
1. Create class Area.
2. Define method calculate().
3. If only one value is provided, calculate the square's area.
4. If two values are provided, calculate the rectangle's area.
5. Return the result.
```

```python
class Area:
    def calculates(self,a=1,b=1):
        if b == 1:
            return a*a

        else:
            return a*b


ar=Area()
print("Area of Square:",ar.calculates(2))
print("Area of Rectangle:",ar.calculates(5,6))
```

```python
## 4. Create a MathOperations class that multiplies any number of values.
Concept
Method Overloading using *args
What to do
Multiply:
• 2 numbers
• 3 numbers
• 4 numbers
• More numbers
How to solve
1. Create class MathOperations.
2. Define method multiply(*args).
3. Initialize the answer as 1.
4. Loop through all numbers.
5. Multiply each number.
6. Return the final product.
```

```python
class MathOperations:
    def multiplie(self,*arg):
        result = 1
        for num in arg:
            result=result * num
        return result
        

math=MathOperations()
print(math.multiplie(1,2,3))
print()
print(math.multiplie(2,3,4))
print()
print(math.multiplie(2,3,4,5,6))
```

```python
## 5. Create a Display class that prints different messages based on the argument type.
Concept
Method Overloading + Type Checking
What to do
Accept:
• Integer
• String
• List
How to solve
1. Create class Display.
2. Create method show().
3. Use isinstance().
4. Print different messages depending on the object's type.
```

```python
class Display:
    def show(self, value):
        if isinstance(value, int):
            print("Integer value:", value)

        elif isinstance(value, str):
            print("String value:", value)

        elif isinstance(value, list):
            print("List value:", value)

        else:
            print("Unknown data type")

obj = Display()

obj.show(100)
print()

obj.show("Hello gokul")
print()

obj.show([1, 2, 3, 4])
    
```

```python
## 6. Create a Volume class that calculates the volume of a cube and cuboid.
Concept
Method Overloading
What to do
• One parameter → Cube
• Three parameters → Cuboid
How to solve
1. Create class Volume.
2. Create method calculate().
3. If one value is given, calculate the cube's volume.
4. Otherwise, calculate the cuboid's volume.
```

```python
class Volume:
    def calculate(self,a,b=None,c=None):
        if b == None and c==None:
            return a **3

        else:
            return a*b*c

obj=Volume()
print(obj.calculate(2))
print(obj.calculate(2,3,5))
        
```

```python
## 7. Create a PrintData class that prints one or many values.
Concept
Method Overloading using *args
What to do
Print:
• One value
• Multiple values
How to solve
1. Create class PrintData.
2. Create method print_data(*args).
3. Loop through the arguments.
4. Print each value.
```

```python
## 11. Create Shape, Circle, and Rectangle classes.
Concept
Method Overriding
What to do
Override the area() method.
How to solve
1. Create Shape.
2. Add area().
3. Inherit Circle.
4. Inherit Rectangle.
5. Override area() in each child class.
```

```python
class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self,r):
        self.Radius=r


    def area(self):
        area=3.14*self.Radius*self.Radius
        print("area of circle:",area)

class Rectangle(Shape):
    def __init__(self,width,height):
        self.Width=width
        self.Height=height

    def area(self):
        area=self.Width * self.Height
        print("area of Rectangle:",area)


circle=Circle(5)
circle.area()
rectangle=Rectangle(89,54)
rectangle.area()

              

    
```

```python
## 12. Create BankAccount, SavingsAccount, and CurrentAccount.
Concept
Method Overriding
What to do
Override the interest() method.
How to solve
1. Create BankAccount.
2. Define interest().
3. Override it in both child classes.
4. Print different interest rates.
```

```python
class BankAccount:
    def __init__(self, name, account_no, balance):
        self.name = name
        self.account_no = account_no
        self.balance = balance

    def interest(self):
        print("Interest rate is not specified.")


class SavingsAccount(BankAccount):
    def interest(self):
        print("Account Holder:",self.name)
        print("Account Number:",self.account_no)
        print("Balance:", self.balance)
        print("Interest Rate :5%")
        print("Interest Amount:",self.balance * 5 / 100)


class CurrentAccount(BankAccount):
    def interest(self):
        print("Account Holder :", self.name)
        print("Account Number :", self.account_no)
        print("Balance:",self.balance)
        print("Interest Rate :2%")
        print("Interest Amount:",self.balance * 2 / 100)


savings = SavingsAccount("gokul", 1001, 50000)
current = CurrentAccount("mukase", 2001, 75000)
print("Savings Account Details")
savings.interest()
print()
print("Current Account Details")
current.interest()
```

```python
## 13. Create Parent and Child classes using super().
Concept
Method Overriding + super()
What to do
Override display().

Sumit Sir Notes
How to solve
1. Create Parent.
2. Add display().
3. Create Child.
4. Override display().
5. Call super().display().
6. Print additional information.
```

```python
class Parent:
    def __init__(self,name,age,):
        self.Name=name
        self.Age=age

    def display(self):
        print("name:",self.Name)
        print("age:",self.Age)


class Child(Parent):
    def __init__(self,name,age):
        super().__init__(name,age)


    def display(self):
        super().display()
        print("name:",self.Name)
        print("age:",self.Age)

p=Parent("rabi",50)
p.display()
child=Child("gokul",21)
child.display()
    
```

```python
## Operator Overloading Questions (15-20)
15. Create a Point class and overload the + operator.
Concept
Operator Overloading (__add__)
What to do
Add two points together.
How to solve
1. Create class Point.
2. Store x and y coordinates.
3. Overload __add__().
4. Return a new point with added coordinates.
```

```python
class Point:
    def __init__(self,x):
        self.X=x
        

    # dunder method:__add__
    def __add__(self,y):
        result=self.X + y.X
        return result

p1=Point(5)
p2=Point(4)
print(p1+p2)
        
```

```python
## 16. Create a Book class and overload the + operator.
Concept
Operator Overloading
What to do
Add the number of pages of two books.
How to solve
1. Create class Book.
2. Store pages.
3. Overload __add__().
4. Return the total pages.
```

```python
class Book:
    def __init__(self,pages):
        self.Pages=pages

    def __add__(self,page):
        result=self.Pages + page.Pages
        return result

b1=Book(200)
b2=Book(100)
print("total book:",b1+b2)
    
```

```python
## 17. Create a Student class and overload the > operator.
Concept
Comparison Operator Overloading (__gt__)
What to do
Compare marks of two students.
How to solve
1. Create class Student.
2. Store marks.
3. Overload __gt__().
4. Compare the marks.
5. Return True or False.
```

```python
class Student:
    def __init__(self,marks):
        self.Marks=marks


    def __gt__(self,mark):
        result=self.Marks > mark.Marks
        return result

s1=Student(input("enter your mark:"))
s2=Student(input("enter your mark:"))
print(s1>s2)
```

```python
## 18. Create a Rectangle class and overload the == operator.

Sumit Sir Notes
Concept
Equality Operator Overloading (__eq__)
What to do
Compare rectangles using their area.
How to solve
1. Create class Rectangle.
2. Store length and width.
3. Calculate area.
4. Overload __eq__().
5. Compare the areas.
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.Length = length
        self.Width = width

    def area(self):
        area = self.Length * self.Width
        return area

    def __eq__(self, other):
        return self.area() == other.area()


r1 = Rectangle(10, 5)
r2 = Rectangle(2, 25)

print(r1 == r2)
```
