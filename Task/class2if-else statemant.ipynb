## 1. Write a program to check whether a number is positive, negative, or zero.

```python
num=int(input("enter your number"))
if num>=0:
     print("number is +ve",num)
elif num<=0:
    print("number is -ve",num)
elif num==0:
    print("number is zero",num)
    
```

## 5. Write a program to check whether a character is uppercase or lowercase.

```python
text=input("enter your character:")

if text>='A'and text<='Z':
    print("the character is uppercase:",text)
elif text>='a'and text<='z':
    print("character is lowercase:",text)
```

## 7. Write a program to display:
o "Morning" if time < 12
o "Afternoon" if time between 12 and 5
o "Evening" otherwise

```python
time=int(input("enter your time:"))

if time<12:
    print("morning")
elif time<12 and time>=5:
    print("afternoon")
else:
    print("evening")

```

## 8. Write a program to check whether a student passed with distinction.
o 75+ → Distinction
o 35–74 → Pass
o Below 35 → Fail

```python
mark=int(input("enter your mark:"))
if mark>=75:
    print("passed with distinction")
elif mark>=35 and mark<=74:
    print("pass")
else:
    print("you fail")
```
