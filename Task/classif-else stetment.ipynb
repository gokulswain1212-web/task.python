## 15. Take three sides of a triangle. Check if they form a valid triangle. If
yes, also print if it is Equilateral, Isosceles, or Scalene.

```python
t1=float(input("enter your tringle side"))
t2=float(input("enter your tringle side"))
t3=float(input("enter your tringle side"))

if t1==t2 and t2==t3 and t3==t1:
    print("Equilateral triangal")
elif t1!=t2 or t2!=t3 or t3!=t1:
    print("Scalene triangal")
elif t1!=t2 or t2!=t3 and t3==t1:
    print("  Isosceles triangal")
```
