# Data Type & Invalid Values
# 11.The Age column contains numbers as well as values like "unknown" and"twenty five". Clean this column and convert it into a proper numeric column.

```python
df['age'] = df['age'].replace({
    "unknown": np.nan,
    "twenty five": 25
})

df['age'] = pd.to_numeric(df['age'], errors="coerce")
```

```python
df['age'].duplicated().sum()
```

## 12.The Annual Income column contains values such as 650000 INR, ₹720000,1.2L, and N/A. Clean the column and convert the valid values into numericformat.

```python
df['annual_income'] = df['annual_income'].replace({
    '650000 INR': 650000,
    '₹720000': 720000,
    '1.2L': 120000,
    'N/A': np.nan
})

df['annual_income'] = pd.to_numeric(df['annual_income'], errors='coerce')

df['annual_income']
```

## 13.Clean the Credit Score column and identify values that cannot be convertedinto numbers.

```python
df['credit_score'] = pd.to_numeric(df['credit_score'], errors='coerce')
```

```python
df['credit_score']
```

## 14.The Loan Amount column contains currency symbols, text, and numericvalues. Remove the unwanted characters and convert it into numeric format.

```python
df['loan_amount_(inr)']=(df['loan_amount_(inr)'].astype(str).str.replace("$","").str.replace(",","").str.replace("INR","").str.strip())
df['loan_amount_(inr)'] = pd.to_numeric(df['loan_amount_(inr)'], errors='coerce')
```

```python
df['loan_amount_(inr)']
```

## 15.The Monthly Visits column contains values such as "five", "N/A", and "-".Convert valid values into numeric values and handle invalid values.

```python
df['monthly_visits']=(df['monthly_visits'].astype(str).str.replace("five","5").str.replace("N/A","").str.replace("_","").str.strip())
df['monthly_visits']=pd.to_numeric(df['monthly_visits'], errors='coerce')
```

```python
df['monthly_visits']
```

## 16.Clean the Debt to Income column where some values are written like 35%and 0.4x. Convert the valid values into a consistent numeric format.

```python
df['debt_to_income']=(df['debt_to_income'].astype(str).str.replace("35%","35").str.replace("0.4x","0.4").str.strip())
df['debt_to_income']=pd.to_numeric(df['debt_to_income'],errors="coerce")
```
