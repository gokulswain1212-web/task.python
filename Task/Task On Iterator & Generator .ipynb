```python
def multiplication_table():
    num=1
    n=5
    while num <= 10:
        yield num * n
        num+=1

mul=multiplication_table()
for i in mul:
    print(i)
```

```python
## Level 5: Advanced Iterator Problems
Problem 21
Create a custom iterator that prints the first 15 natural numbers.
```

```python
class NaturalNumberIterator:
    def __init__(self):
        self.number = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.number <= 15:
            value = self.number
            self.number += 1
            return value
        else:
            raise StopIteration

natural_numbers = NaturalNumberIterator()

for num in natural_numbers:
    print(num)
```

```python
## Problem 23
Create a custom iterator that prints all alphabet letters from A to Z.
```

```python
class alphabet_letters:
    def __init__(self):
        self.Letters=ord("A")

    def __iter__(self):
        return self

    def __next__(self):
        if self.Letters <=ord ("Z"):
            value=chr(self.Letters)
            self.Letters+=1
            return value
        else:
            raise StopIteration

alphabet=alphabet_letters()
for i in alphabet:
    print(i)
```

```python
## Problem 24
Create a custom iterator that prints all vowels from the string below.
text = "Artificial Intelligence"
```

```python
class Vowels:
    def __init__(self, text):
        self.text = text
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        while self.index < len(self.text):
            char = self.text[self.index]
            self.index += 1

            if char.lower() in "aeiou":
                return char

        raise StopIteration


text = "Artificial Intelligence"

vowel_iterator = Vowels(text)

for vowel in vowel_iterator:
    print(vowel)
```
