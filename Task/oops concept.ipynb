```python
class Student:
    # instance method
    def Student_info(self,name,roll_no,class_n):
        self.name=name
        self.roll_no=roll_no
        self.class_n=class_n


    # method..2:
    def display_information(self):
        print(f"Student name:{self.name}")
        print(f"Student roll_no:{self.roll_no}")
        print(f"Student class_n:{self.class_n}")


# creat the class object

std1 =Student()

# call the method 1:

#student_info
std1.Student_info("gokul",86,"nareshit")

# call thr mrthod2:

std1.display_information()
```
