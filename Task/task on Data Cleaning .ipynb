## 9. Unit Price Cleaning
## The unit_price column contains values such as ₹15000, $20000, 25000 INR, and 10k.Convert all valid values into a single numeric format.

```python
df['unit_price']=df['unit_price'].astype(str).str.replace("₹","").str.replace("$","").str.replace("INR","").str.replace("k","")
df['unit_price']=pd.to_numeric(df['unit_price'],errors="coerce")
df['unit_price']
df['unit_price']=df['unit_price'].astype('float')
df['unit_price']
```

## 10. Unit Price Outliers
# Detect extreme unit_price values using the IQR method. Compare the number ofoutliers before and after treatment.

```python
Q1 = df['unit_price'].quantile(0.25)
Q3 = df['unit_price'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers_before = ((df['unit_price'] < lower) | 
                   (df['unit_price'] > upper)).sum()

median_price = df['unit_price'].median()
df.loc[(df['unit_price'] < lower) | 
       (df['unit_price'] > upper), 'unit_price'] = median_price

outliers_after = ((df['unit_price'] < lower) | 
                  (df['unit_price'] > upper)).sum()

print("Outliers before:", outliers_before)
print("Outliers after:", outliers_after)
```
