```python
## 1. Person → Student, Teacher
Question
Create a parent class Person with attributes name and age. Create two child classes Student
and Teacher that inherit from Person.
Concept
• Hierarchical Inheritance
• Parent attributes inherited by multiple child classes.
How to Solve
1. Create a parent class Person.
2. Add name and age using a constructor.
3. Create Student(Person).
4. Create Teacher(Person).
5. Create objects of both classes.
6. Show that both children can access name and age.
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

class Student(Person):
    def __init__(self,name,age,subject):
        super().__init__(name,age)
        self.Student_subject=subject

    def display_student_info(self):
        print("student name:",self.name)
        print("student age:",self.age)
        print("subject:",self.Student_subject)

    

class Teacher(Person):
    def __init__(self,name,age,school_name):
        super().__init__(name,age)
        self.School_name=school_name

    def display_Teacher_info(self):
        print("Teacher name:",self.name)
        print("Teacher age:",self.age)
        print("school_name:",self.School_name)


s=Student("gokul",21,"python")
s.display_student_info()

t=Teacher("raju",40,"SSB-school")
t.display_Teacher_info()


        
```

```python
## 4. Shape → Circle, Rectangle
Question
Define class Shape with method area(). Create subclasses Circle and Rectangle.
Concept
• Parent provides common method.
• Children may later override it.
How to Solve
1. Create Shape.
2. Add area().
3. Create Circle.
4. Create Rectangle.
5. Call area().
```

```python
## 16. Shape Area
Concept
Polymorphism using overriding.
How to Solve
1. Parent constructor.
2. Circle calculates πr2.
3. Square calculates side2.
4. Override area().
```

```python
class Shape:
    def area(self):
        print("Area of shape")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        area = 3.14 * self.radius * self.radius
        print("Circle Radius:", self.radius)
        print("Circle Area:", area)

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        area = self.side * self.side
        print("Square Side:", self.side)
        print("Square Area:", area)

circle = Circle(7)
square = Square(5)

circle.area()
print()
square.area()
```

```python
## 19. Device Info
Concept
Method overriding.
How to Solve
1. Parent method info().
2. Phone overrides.
3. Tablet overrides.
```

```python
## 20. Game Start
Concept
Inheritance + Additional Methods.
How to Solve
1. Parent method start().
2. Chess inherits.
3. Chess adds checkmate().
4. Ludo adds roll_dice().
```

```python

class Game:
    def start(self):
        print("Game started")

class Chess(Game):
    def checkmate(self):
        print("Checkmate move completed in Chess")

class Ludo(Game):
    def roll_dice(self):
        print("Dice rolled in Ludo")

chess_game = Chess()
ludo_game = Ludo()

chess_game.start()
ludo_game.start()
chess_game.checkmate()
ludo_game.roll_dice()
```
