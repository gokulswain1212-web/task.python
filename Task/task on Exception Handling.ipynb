## 21. Create a custom exception InsufficientBalanceError for a banking program.

```python
class InsufficientBalanceError(Exception):
    pass

try:
    class Bank:
        def __init__(self,balance):
            self.Balance=balance

        def withdrawl(self,amt):
            if amt  <= self.Balance:
                print("withdrawl balance")

            else:
                raise InsufficientBalanceError("insufficientbalance")

    obj=Bank(500)
    obj.withdrawl(600)

except InsufficientBalanceError as e:
    print(e)
```

## 22. Write a program that raises TypeError manually if the input is not a string.

```python
class TypeError(Exception):
    pass

try:
    value = 123

    if not isinstance(value, str):
        raise TypeError("Input must be a string.")

    print(value)

except TypeError as e:
    print(e)
    
```
